```python
import sys
sys.path.append("..")

from pathlib import Path
import numpy as np
import matplotlib.pyplot as plt
from PIL import Image

DATA_RAW       = Path("../data/raw")
DATA_PROCESSED = Path("../data/processed")

print("Setup abgeschlossen ✓")
```

```python
infected_raw = list((DATA_RAW / "infected").glob("*.*"))
healthy_raw  = list((DATA_RAW / "healthy") .glob("*.*"))

print(f"Infected : {len(infected_raw):,}")
print(f"Healthy  : {len(healthy_raw):,}")
print(f"Gesamt   : {len(infected_raw) + len(healthy_raw):,}")
```

```python
fig, axes = plt.subplots(2, 5, figsize=(12, 5))

for i, img_path in enumerate(infected_raw[:5]):
    axes[0, i].imshow(Image.open(img_path).convert("RGB"))
    axes[0, i].axis("off")
    axes[0, i].set_title("Infected" if i == 0 else "")

for i, img_path in enumerate(healthy_raw[:5]):
    axes[1, i].imshow(Image.open(img_path).convert("RGB"))
    axes[1, i].axis("off")
    axes[1, i].set_title("Healthy" if i == 0 else "")

plt.tight_layout()
plt.show()
```
